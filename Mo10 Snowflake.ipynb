### A.

I collected the data by pulling daily “top gainers” lists from two sources: the Wall Street Journal and Yahoo Finance. They were individually saved as timestamped CSVs and then loaded into Snowflake. All of these raw tables live in the DS2508 database, JCH7AA schema. In total there are 22 WSJGAINERS and 21 YGAINERS tables. After loading, I used a Python normalization step to standardize both sources into the same column structure (e.g., symbol, company name, last price, daily and percent change, volume), so that all of the gainers data can be treated as one consistent dataset for models.
